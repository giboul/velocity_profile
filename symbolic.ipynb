```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

def display_eq(a, b):
    display(sp.Equality(a, b))
```

```python
z = sp.Symbol("z", real=True)

u = sp.Function("u")(z)
e = sp.Function("e")(z)
zls = sp.Function("Z_LS")(z)
tv = sp.Function("tau_v")(z)
tt = sp.Function("tau_t")(z)
td = sp.Function("tau_d")(z)
f = sp.Function("f")(z)

du = u.diff(z)
d2u = u.diff(z, 2)
de = e.diff(z)
d2e = e.diff(z, 2)
dzls = zls.diff(z)

rho, lam, eb, dp, k, nu, a, ae, be, g, i = sp.symbols(r"rho lambda epsilon_b d_p kappa nu A_\ast A_E B_E g i_s")
```

```python
tv_expression = rho * nu * (e * u).diff(z)
display_eq(tv, tv_expression)
display_eq(tv.diff(z), tv_expression.diff(z))
```

```python
phi = - sp.sqrt(zls*u/nu)/a
tt_expression = rho * e * k**2 * zls**2 * (1-sp.exp(phi))**2 * du**2
display_eq(tt, tt_expression)
display_eq(tt.diff(z), tt_expression.diff(z))
```

```python
td_expression = rho * lam * dp / (1-eb) * (1-e) * e * u * du
display_eq(td, td_expression)
display_eq(td.diff(z), td_expression.diff(z))
```

```python
f_expression = (
    - ae * (1-e)**2 / (e * dp**2) * rho * nu * u
    - be * (1-e) * e / dp * rho * u**2
)
display_eq(f, f_expression)
```

```python
governing_equation = sp.Equality(0, e*rho*g*i + (tt + tv + td).diff(z) + e*f)
display(governing_equation)
governing_equation = e*rho*g*i + tv_expression.diff(z).expand() + tt_expression.diff(z) + td_expression.diff(z) + e*f_expression
display_eq(0, governing_equation)
```

```python
q = governing_equation.subs({d2u: 0})
p = (governing_equation - q).subs({d2u: 1})
display_eq(sp.Symbol("p"), p)
display_eq(sp.Symbol("q"), q)
display((p*d2u + q - governing_equation).expand())
```

```python
constants = {
    rho: 950,
    lam: 0.015,
    eb: 0.38,
    dp: 8e-3,
    k: 0.41,
    nu: 3.1e-6,
    a: -2.0 / dp * np.log(1.0 / 0.99 - 1.0),
    ae: 180.,
    be: 1.75,
    g: 9.81,
    i: 1e-2,
}

functions = {
    e: eb + (1.0 - eb) * 0.5 * (1.0 + sp.tanh(0.5 * a * z)),
    zls: sp.log(1.0 + sp.exp(a * z)) / a
}
```

```python
d2u_expr = - q / p
d2u_func = sp.lambdify([z, u, du], d2u_expr.subs(functions).subs(constants).doit(), cse=True)
d2u_func(-10., 4e-3, 0.)

jac = sp.Matrix([
    [0, 1],
    [d2u_expr.diff(u), d2u_expr.diff(du)]
])

display(jac)

jac_func = sp.lambdify([z, u, du], jac.subs(functions).subs(constants).doit(), cse=True)
jac_func(-1., 4e-3, 0.)
```
